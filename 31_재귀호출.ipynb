```python
# 재귀함수 - 자기 자신을 호출하는 방식

# 재귀호출 사용하기

def hello() :
    print("hello, world")
    hello()
```

```python
# 재귀호출에 종료 조건 만들기

def hello(count) :
    if count == 0 :
        return

    print("Hello world", count)

    count -= 1
    hello(count)

hello(5)
```

```python
# 재귀호출로 팩토리얼 구하기

def factorial(n) :
    if n == 1 :
        return 1
    return n * factorial(n - 1)

factorial(5)
```

```python
# 연습문제 31.4

# 회문 판별

def is_palindrome(word) :
    if len(word) < 2 :
        return True
    if word[0] != word[-1] :
        return False
    return is_palindrome(word[1:-1])

print(is_palindrome('hello'))
print(is_palindrome('level'))   
```

```python
# 심사문제 31.5

def fib(num1) :
    if num1 == 0 :
        return 0
    
    elif num1 == 1:
        return 1
    
    else :
        return fib(num1 - 1) + fib(num1 - 2)
    
n = int(input())
print(fib(n))
```
