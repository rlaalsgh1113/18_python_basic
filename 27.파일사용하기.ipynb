```python
# 파일에 문자열 쓰기 읽기

## 파일에 문자열 쓰기

# 파일객체 = open(파일이름, 파일모드)    
# 파일객체.write('문자열')
# 파일객체.close()

file = open("hello.txt", "w")
file.write("hello world")
file.close()
```

```python
# 파일에서 문자열 읽기
# 변수 = 파일객체.read()

file = open("./data/hello.txt", "r")
s = file.read()
print(s)
file.close()
```

```python
# 자동으로 파일 객체 닫기

# with open("파일이름", "파일모드") as 파일객체 :
    # 코드
# 내어 쓰기 하게 되면 파일 자동으로 닫힘

with open("./data/hello.txt", "r") as file :
    s = file.read()
    print(s)
```

```python
# 문자열 여러 줄을파일에 쓰기, 읽기

with open("./data/hello.txt", "w") as file :
    for i in range(3) :
        file.write(f"Hello, world! {i}, i\n")
```

```python
## 파일 객체.writelines(문자열리스트)
lines = ["안녕하세요.\n", "파이썬\n", "코딩도장입니다.\n"]

with open("./data/hello.txt", "w") as file :
    file.writelines(lines)
```

```python
# 한줄씩 리스트로 가져오기

with open("./data/hello.txt", "r") as file :
    lines = file.readlines()
    print(lines)
```

```python
# 파일의 내용을 한줄씩 읽기

with open("./data/hello.txt", "r") as file :
    line = None
    while line != "" :
        line = file.readline()
        print(line.strip("\n"))
```
