```python
# 중첩 루프 사용하기

for i in range(5) : # 바깥 루프는 세로
    for j in range(5) : # 안 루프는 가로
        print("J:", j, sep="",end=" ")
    print("i:", i, "\\n", sep="")
```

```python

# 사각형으로 별 출력하기
for i in range(5) :
    for j in range(5) :
        print("*",end="")
    print()
```

```python
# 사각형 모양 바꾸기

for i in range(3) :
    for j in range(7) :
        print("*", end="")
    print() # 기본값 \n
    
```

```python
# 계단식으로 별 출력하기

for i in range(5) :
    for j in range(5) :
        if j <= i :
            print("*", end="")
    print()
```

```python
# 대각선으로 별 출력하기

for y in range(5) :
    for x in range(5) :
        if y == x :
            print("*",end = "")
        else :
            print(" ", end = "")
    print()
```

```python

# 연습문제 19.5

for i in range(5) :
    for j in range(5) :
        if j >= i :
            print("*", end="")
        else :
            print(" ", end="")
    print()        
```

```python
# 심사문제 19.6

a = int(input())

for i in range(a) :
    for j in range(a+a-1) :
        if j <= i :
            print("*", end="")
        elif j > i :
            print(" ",end="")
    print()
```

```python
# 심사문제 19.6

a = int(input())

for i in range(a):
    for j in range(a - i - 1):
        print(" ", end="")
        
    for j in range(2 * i + 1):
        print("*", end="")
        
    print()
```
