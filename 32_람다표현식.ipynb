```python
# 람다표현식 자체를 호출하기
# (lambda 매개변수 : 식)(인수들(arg))

(lambda x : x + 10)(2)
                 #  ^ 매개변수
```

```python
# 람다표현식 안에서는 변수를 만들 수 없다.

(lambda x : y = 10; x + y)(2)
```

```python
# 람다 표현식과 map, filter, reduce 함수 활용

a = list(range(1, 11))

list(map(lambda x : str(x) if x % 3 == 0 else x, a))
                  # 조건 만족시 if 앞 아닐 시 else 뒤
                  # 람다 표현식 if 사용 시 else 필수
```

```python
# map에 객체 여러개 넣기

a = [1, 2, 3, 4, 5]
b = [2, 4, 6, 8, 10]

list(map(lambda x, y : x * y, a, b))

```

```python
# filter 사용하기
# filter(함수, 반복가능한 객체)

def f(x) : 
    return 5 < x < 10

a = [8, 3, 2, 10, 15, 7, 1, 9, 0, 11]

list(filter(f, a))
```

```python
a = [8, 3, 2, 10, 15, 7, 1, 9, 0, 11]

list(filter(lambda x : 5 < x < 10, a))
```

```python
# reduce 사용하기

# import random
# random.randint(1,10)

# from random import randint
# from functools import reduce

from functools import reduce
# reduce(함수, 반복가능객체)

def f(x,y) :
    return x + y

f(1, 2)

a = [1, 2, 3, 4, 5]
reduce(f, a)
# reduce란 누적해서 더해주는 함수
```

```python
a = [1, 2, 3, 4, 5]

    # 모듈            # 클래스나 함수나 변수
from functools import reduce

reduce(lambda x, y : x + y, a)
```

```python
# 연습문제 32.4
# p.417

files = ['font', '1.png', '10.png', '11.gif', '2.jpg', '3.png', 'table.xslx', 'spec.docx']

list(filter(lambda x : x.find('.jpg') != -1  or x.find('.png') != -1, files))
```
