```python
# 딕셔너리 만들기
# 딕셔너리 = {
#     키1 : 값1,
#     키2 : 값2,
#     키3 : 값3
# }
```

```python
lux = {
    'health' : 490,
    'mana' : 334,
    'melee' : 550,
    'armor' : 18.72
}
print(lux)
```

```python
# 키 이름이 중복이면?

lux = {
    'health' : 490,
    'health' : 800,   # 키가 중복이면 가장 뒤에 있는 값만 사용
    'mana' : 334,
    'melee' : 550,
    'armor' : 18.72
}
print(lux["health"])
```

```python
# 딕셔너리의 키의 자료형
# 키에는 리스트, 딕셔너리 사용 불가 - 리스트와 딕셔너리는 변경 가능한 값이기때문

x = {
    100 : "hundred",
    False : 0,
    3.5 : [3.5, 3.5]
}
print(x)
```

```python
x = {
    [10, 20] : 100
}
```

```python
# 빈 딕셔너리 만들기

딕셔너리 = {}
딕셔너리 = dict()

# int(), str(), float(), list(), tuple(), bool()

x = {}
y = dict()

print(type(x))
print(type(y))
```

```python
# dict 로 딕셔너리 만들기
lux1 = dict(
    health=490, 
    mana=334, 
    melee=550, 
    armor=18.72)
print(lux1)
```

```python
iter1 = ["health", "mana", "melee", "armor"]  
iter2 = [490, 334, 550, 18.72]
dict(zip(iter1, iter2))   # zip - 결합시켜주는 함수
```

```python
lux3 = dict(
    [("health", 490), ("mana", 334), ("melee", 550), ("armor", 18.72)]
)
print(lux3)
```

```python
lux4 = dict(
    {'health': 490, 'mana': 334, 'melee': 550, 'armor': 18.72}
)
print(lux4)
```

```python
# 딕셔너리의 키에 접근하고 값 할당하기

lux = {
    'health' : 490,
    'mana' : 334,
    'melee' : 550,
    'armor' : 18.72
}
print(lux['health'])
print(lux['melee'])
```

```python
# 딕셔너리 키의 값 할당

lux = {
    'health' : 490,
    'mana' : 334,
    'melee' : 550,
    'armor' : 18.72
}
lux['health'] = 2037
lux['mana'] = 1184

print(lux)
```

```python
# 없는 키 값 적용하기
lux['mana_ragen'] = 3.28
print(lux)
```
