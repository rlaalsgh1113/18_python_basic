```python
class DutchPayManager:
    def calculate_dutch_pay(self ,total_amount, people_count):
        pay_per_person = total_amount / people_count
        return int(pay_per_person)

price = int(input())
persons = int(input())

manager = DutchPayManager()

my_pay = manager.calculate_dutch_pay(price, persons)

print(f"한 사람당 내야 할 금액은 {my_pay}원입니다.")
```

```python
x = 10
y = 3

def get_quotient_remainder(x,y) :
    return x // y , x % y
quotient, remainder = get_quotient_remainder(x, y)
print("몫: {0}, 나머지: {1}".format(quotient, remainder))
```

```python
x, y = map(int, input().split())

def calc(a, b) :
    return a + b, a - b, a * b, a / b
    
a, s, m, d = calc(x,y)
print("덧셈 : {0}, 뺄셈 : {1}, 곱셈: {2}, 나눗셈 : {3}".format(a, s, m, d))
```
