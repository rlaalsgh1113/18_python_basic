```python
class DutchPayManager:
    def calculate_dutch_pay(self ,total_amount, people_count):
        pay_per_person = total_amount / people_count
        return int(pay_per_person)

price = int(input())
persons = int(input())

manager = DutchPayManager()

my_pay = manager.calculate_dutch_pay(price, persons)

print(f"한 사람당 내야 할 금액은 {my_pay}원입니다.")
```

```python
menu = {
    "coffee" : 4500,
    "tea" : 3500,
    "latte" : 5000
}

sale_record = []

def calc(date, order) :
    if order in menu :
        price = menu[order]
        print(f"{price}원 입니다.")

        sale_record.append({"date" : date, "price" : price})

    else :
        print("해당 메뉴는 없습니다.")

def sales(target_date) :
    total_sale = 0

    for record in sale_record:
        if target_date in record["date"] :
            total_sale += record["price"]
    print(f"[{target_date}] 매출 : {total_sale}원")

date1 = input("날짜 입력 ex)2026-09")
order = input("메뉴 입력")

calc(date1, order)

sales(date1)
```
