```python
# 세트  만들기

# 세트 = {값1, 값2, 값3}
## 1. 순서가 없음  -- > 인덱싱, 슬라이싱 허용하지 않음
## 2. 중복을 허용하지 않음

fruits = {
    'strawberry', 'grape', 'orange', 'pineapple', 'cherry'
}
print(fruits)
```

```python
fruits = {
    "orange", "orange", "cherry"
}
print(fruits)
```

```python
# set 함수를 사용하여 세트 만들기

a = set("apple")

print(type(a))
print(a)
```

```python
b = set(range(5))
print(b)
```

```python
c = set()
print(c)

# 출력값이 {} = > 딕셔너리 형태
```

```python
# 집합 연산 사용하기

## 합집합
## 세트1 | 세트2
## set.union(세트1, 세트2) => 세트1.union(세트2)

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a | b)
print(set.union(a, b))
print(a.union(b))
```

```python
# 교집합
## 세트1 & 세트2
## set.intersection(a, b)
## a.intersection(b)

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print(a & b)
print(set.intersection(a, b))
print(a.intersection(b))
```

```python
# 차집합
## 세트1 - 세트2
## a - b
## set.difference(a, b)
## a.difference(b)

a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print(a - b)
print(set.difference(a, b))
print(a.difference(b))
```

```python
# 대칭차집합
## 세트1 ^ 세트2
## set.symmetric_difference(a, b)

print(a ^ b)  # 교집합 반대
print(set.symmetric_difference(a, b))
```

```python
# 집합 연산 후 할당 연산자 사용하기
## 세트1 |= 세트2
## 세트1.update(세트2)

a = {1, 2, 3, 4}

a |= {5}
print(a)

a = {1, 2, 3, 4}
a.update({5})
print(a)
```

```python
# intersection_update
## 세트1 &= 세트2
## 세트1.intersection_update(세트2)

a = {1, 2, 3, 4}
a &= {0, 1, 2, 3, 4}
print(a)

a = {1, 2, 3, 4}
a.intersection_update({0, 1, 2, 3, 4})
print(a)
```

```python
a = {1, 2, 3, 4}
a ^= {3, 4, 5, 6}
print(a)
```

```python
# 부분집합과 상위집합 확인하기

## 현재 세트 <= 다른세트
## 현재 세트.issubset(다른세트)

a = {1, 2, 3, 4}

print(a <= {1, 2, 3, 4})
a.issubset({1, 2, 3, 4})
```

```python
a = {1, 2, 3, 4}
a < {1, 2, 3, 4, 5, 6}
```

```python
# 세트가 같은지 다른지 확인하기

a = {1, 2, 3, 4}
a == {1, 2, 3, 4}
```

```python
a == {4, 1, 2, 3}
```

```python
# 세트 조작하기
## 세트에 요소 추가하기

a = {1, 2, 3, 4}
a.add(5)
print(a)
```

```python
# 세트에 요소 삭제하기

a.remove(3)
print(a)
```

```python
a.discard(2) # discard는 요소가 없으면 에러가 뜨지 않고 넘어감
print(a) 
```

```python
# 세트에서 임의의 요소 삭제하기

a = {1, 2, 3, 4}
print(a.pop())
print(a)
```

```python
# 세트의 모든 요소 삭제하기

a = {1, 2, 3, 4}
a.clear()
print(a)
```

```python
# 반복문으로 set 요소 모두 출력하기
for i in {1, 2, 3, 4, 5} :
    print(i)
```

```python
# 세트 표현식 사용하기
## {식 for 변수 in 반복 가능한 객체}
## set(식 for 변수 in 반복 가능한 객체)

a = {i for i in "apple"} # 세트와 딕셔너리 구분하기 딕셔너리는 i 뒤에 : 나옴
print(a)
```

```python
a = {i for i in "pineapple" if i not in "apl"}
print(a)
```

```python
# 연습문제 26.8

a = {i for i in range(1, 101) if i % 3 == 0}
b = {i for i in range(1, 101) if i % 5 == 0}

print(a & b)
```

```python
# 심시문제 26.9

num1, num2 = map(int, input().split())

a = {i for i in range(1, num1 + 1) if num1 % i == 0}
b = {i for i in range(1, num2 + 1) if num2 % i == 0}

print(a)
print(b)

divisor = a & b

result = 0
if type(divisor) == set :
    result = sum(divisor)

print(result)
```
