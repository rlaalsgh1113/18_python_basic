```python
# 문자열 서식 지정자와 포매팅 사용하기


# 서식 지정자로 문자열 넣기
# %s % "문자열"

"I am %s" % "james"
```

```python
name = "maria"
"I am %s" %name
```

```python
# 서식지정자로 숫자 넣기

# "%d" % 숫자

"I am %d" %26

b = 10

"I am %d" %b
```

```python
# 서식지정자로 소수점 표현하기

# "%f" % 숫자(실수)

"%f" % 2.3
```

```python
# "%.자리수f" % 숫자

"%.2f" % 2.3156    # % 뒤에 들어오는 숫자는 공간 확보 ex %5.2f   " 2.32 "
```

```python
# 서식 지정자로 문자열 정렬하기
# %길이s

"%10s" % "python"
```

```python
# 왼쪽 정렬하기

"%-10s" % "python"
```

```python
# 서식 지정자로 문자열 안에 값 여러개 넣기

# "%d %s" % (숫자, 문자열)

"Today is %d %s" % (3, "April")
```

```python
# format 매서드 사용하기

print("Hello, {0}".format("world"))
```

```python
# format 매서드로 값을 여러개 넣기

print("Hello, {0} {2} {1}".format("python", "Script", 3.6))
                                     # 0       1        2
```

```python
# format 매서드로 같은 값 여러개 넣기

print("{0} {0} {0} {1}".format("python", "script"))
```

```python
# format 매서드에서 index 생략하기

print("Hello {} {} {}".format("python", "Script", 3.6))
```

```python
# format 매서드에서 인덱스 대신 이름 지정하기

print("Hello {language} {version}".format(language="Python", version=3.6))
```
