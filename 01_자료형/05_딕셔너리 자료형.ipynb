```python
# 딕셔너리는 어떻게 만들까? 
# {key1 : value1, key2 : value2, ...}

dict = {
    "name" : "kim",
    "phone" : "010-1234-5678",
    "birth" : "09-02" 
}

print(type(dict))
```

```python
## 딕셔너리 쌍 추가하기
a = {1 : "a"}
a[2] = "b"
print(a)
```

```python
a["name"] = "김민호"
print(a)
```

```python
a[3] = [1, 2, 3]
print(a)
```

```python
a = {
    1 : "a",
    2 : "b", 
    "name" : "김민호",
    3 : [1, 2, 3],
    (1, 2) : (1, 2),
    # [1, 2] : [1, 2] 딕셔너리에선 리스트가 키값이 될 수 없다
}
print(a)
```

```python
## 딕셔너리에서 key를 사용해서 value값 얻기
학점 = {
    "홍길동" : 10,
    "유재석" : 90
}

print(학점["유재석"])
print(학점["홍길동"])

학점["홍길동"] = 80
print(학점["홍길동"])
학점["강호동"] = 70  
print(학점)
```

```python
## 딕셔너리 관련 함수
# key 리스트 만들기 - keys()
dict = {
    "name" : "kim",
    "phone" : "010-1234-5678",
    "birth" : "09-02" 
}

print(dict.keys())
print(list(dict.keys()))
```

```python
## values 리스트 만들기
dict = {
    "name" : "kim",
    "phone" : "010-1234-5678",
    "birth" : "09-02" 
}
print(dict.values())
print(list(dict.values()))
```
