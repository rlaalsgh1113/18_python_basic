```python
# 문자열은 어떻게 만들고 사용할까?

## 1. 큰 따옴표
print("Hello World")

## 2. 작은 따옴표
print('Python is fun')

## 3. 큰 따옴표 3개 연속으로 감싸기
print("""Life is too short, You need Python""")

## 4. 작은 따옴표 3개 연속으로 감싸기
print('''Life is too short, You need Python''')
```

```python
## 문자열에 큰따옴표 표현하기
print('"python is very fun" chulsu says')
```

```python
## 문자열 길이 구하기 len()

a = "life is too short"
len(a) #len("life is too short")
```

```python
## 문자열 인덱싱과 슬라이싱
a = "life is too short" # 처음 글자 0으로 인식
a[8]
a[-1]
a[-2]
```

```python
## 슬라이싱
a[0:3] # @+1을 해야 원하는 글자까지 출력 가능
a[12:17]
a[-5:-1] 
a[-5: ] # 끝부분 생략시 끝까지 출력
a[:4] # 첫 부분 생략시 처음부터 출력
```
