```python
## 문자열 길이 구하기 len()

a = "life is too short"
len(a) #len("life is too short")
```

```python
## 문자열 인덱싱과 슬라이싱
a = "life is too short" # 처음 글자 0으로 인식
a[8]
a[-1]
a[-2]
```

```python
## 슬라이싱
a[0:3] # @+1을 해야 원하는 글자까지 출력 가능
a[12:17]
a[-5:-1] 
a[-5: ] # 끝부분 생략시 끝까지 출력
a[:4] # 첫 부분 생략시 처음부터 출력
```

```python
## 문자열 포매팅
print("I eat %d apples" % 3)
```

```python
print("I eat %s apples" % "five")
```

```python
a = 3
b = "five"
print("I eat %d apples" % a)
print("I eat %s apples" % b)
```

```python
## format() 함수를 사용한 포매팅
"I eat {0} apples".format(3)
"I eat {0} apples".format("five")
```

```python
## 2개 이상의 값을 넣을때
number = 10
day = "three"
"I ate {0} apples, so i was sick for {1} days".format(number, day)
```

```python
## f문자열 포매팅
name = "홍길동"
age = 30

print(type(name)) # type = 상태
print(type(age))

print(f"나의 이름은 {name}이고, 나이는 {age}입니다.")
```

```python
## 문자열 관련 함수들
a = "hobby"
type(a)

## 특정 문자 개수 세기 - count
a.count("b") 
"hobby".count("o")
```

```python
## 위치 알려주기 - find()
a = "Python is the best choice"
print(a.find("b"))
print(a.find("k")) # 없는 값은 -1로 출력
```

```python
## 위치 알려주기 - index()
a = "Life is too short"
print(a.index("t"))
print(a.index("k")) # index는 없는 값은 에러
```

```python
## 문자열 삽입 - join
",".join(["a", "b", "c", "d"]) # 하나의 문자열로 만듦
```

```python
## 소문자를 대문자로 바꾸기 - upper
a = "hi"
a.upper()

## 대문자를 소문자로 - lower
a = "HI"
a.lower()
```

```python
## 양쪽 공백 지우기 - strip
a = "   hi   "
a.strip()
```

```python
## 문자열 바꾸기 - replace
a = "Life is too short"
a.replace("Life", "Leg")
```

```python
## 문자열 나누기 - split
a = "Life is too short"
a.split()
```
