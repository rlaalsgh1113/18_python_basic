```python
# 리스트는 어떻게 만들고 사용할까?
# 리스트명 = [요소1, 요소2, ... ]

a = []
b = [1, 2, 3]
c = ["Life", "is", "too", "short"]
d = [1, 2, ["Life", "is"]]
e = [1, 2, "Life", "is"]
f = list()

print(a)
print(b)
print(c)
print(d)
print(e)
print(f)
```

```python
## 리스트에 인덱싱
a = [1, 2, 3]
type(a)
a[1]
```

```python
a[2]
a[-1]
```

```python
a = [1, 2, 3, ["a", "b", "c"]]
print(len(a))
print(a[3])
print(a[3][1])
```

```python
## 리스트에 슬라이싱
a = [1, 2, 3, 4, 5]
print(a[1:4])
```

```python
a = [1, 2, 3, ["a", "b", "c"]]
a[-1][1:3] 
```

```python
## 리스트 연산
a = [1, 2, 3]
b = [4, 5, 6]
print(a + b)
print(a * 3)
```

```python
## 리스트 길이 구하기 - len
a = [1, 2, 3]
len(a) 
```

```python
## 리스트 수정과 삭제
a = [1, 2, 3]
a[2] = 4 # 수정
print(a)
```

```python
a = [1, 2, 3]
del a[1] # 삭제 del
print(a)
```

```python
## 리스트 관련 함수

## 리스트에 요소 추가하기 - append
a = [1, 2, 3]
a.append(4)
print(a)
```

```python
a.append(["a","b"])
print(a)
```

```python
## 리스트 정렬
a = [1, 4, 3, 2]
a.sort()
print(a)
```

```python
a = ["a", "b", "c"]
a.sort(reverse=True) # 내림차순 reverse(파라메타)
print(a)
```

```python
## 리스트 뒤집기 - reverse(함수 이름)
a = ["a", "c", "b"]
a.reverse()
print(a)
```

```python
## 인덱스 반환 - index
a = [1, 2, 3, 4]
a.index(3)
```

```python
## 리스트의 요소 삽입 - insert
a = [1, 2, 3]
a.insert(0,4)
print(a)
a.insert(-1,10)
print(a)
```

```python
## 리스트에 요소 제거 - remove()   ()안의 값 지우기
a = [1, 2, 3, 1, 2, 3]
a.remove(2)
print(a)
a.remove(2)
print(a)
```

```python
## 리스트 요소 꺼내기   - pop

a = [1, 2, 3]
number = a.pop()  # 기본값 -1 위치에 해당하는 값이 출력
print(number)
print(a)
```

```python
## 리스트에 포함된 요소 x의 개수 세기 - count
a = [1, 2, 3, 4, 5, 3]
print(a.count(3))
```

```python
## 리스트 확장
a = [1, 2, 3]
a.extend([4, 5])
print(a)
```
