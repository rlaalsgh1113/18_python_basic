```python
# 튜플은 어떻게 만들까? 

t1 = ()
t2 = (1,)
t3 = (1,2,3)
t4 = 1,2,3       # 튜플은 괄호 생략 가능
t5 = ("a", "b", ("ab", "cd"))
t6 = tuple()

type(t3)
```

```python
t1 = (10, )
t2 = 10,
print(type(t1))
print(type(t2))

d1 = (10) # 콤마 없을 시 d1 = 10 과 같음 튜플이 아님
print(type(d1))

num = 10, 20
print(type(num))
```

```python
## 튜플의 요소 값을 지우거나 변경하려고 하면? 
t1 = (1,2,"a","b")
del t1[2]         # 튜플은 지우거나 변경이 되지 않는다
```

```python
t1 = (1,2,"a","b")
t1[2] = "c"
```

```python
##  튜플 인덱싱
t1 = (1,2,"a","b")
print(t1[0])
print(t1[-1])
```

```python
## 튜플 슬라이싱
t1 = (1,2,"a","b")
print(t1[1:])
```

```python
## 튜플 더하기
t1 = (1,2,"a","b")
t2 = (3, 4)
print(t1+t2)
```

```python
## 튜플 곱하기
t1 = (1,2,"a","b")
t2 = (3, 4)
print(t2 * 3)
```
