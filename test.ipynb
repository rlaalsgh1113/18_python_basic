```python
library = {
    "문학" : [
        {"제목" : "1984", "저자" : "조지 오웰", "재고" : 3},
        {"제목" : "위대한 개츠비", "저자" : "F. 스콧 피츠제럴드", "재고" : 0}
    ],
    "과학" : [
        {"제목" : "코스모스", "저자" : "칼 세이건", "재고" : 5},
        {"제목" : "이기적 유전자", "저자" : "리처드 도킨스", "재고" : 2}
    ]
}

total = 0

for gerne, books in library.items() :
    for book in books :

        total += book["재고"]

        if book ["재고"] == 0 :
            print(f"분야 : {gerne}, 제목 : {book["제목"]}")

print(f"{total}")
```

```python
# 9번

a = list(map(int, input().split()))
count = 0
average = sum(a)/len(a)

def analyze_scores(*args) :
    global count

    for score in a :
        if score >= average :
            count += 1

analyze_scores()

print(f"평균 : {average} 평균 이상인 수의 개수 : {count}")

```

```python
# 10번

inventory = {
    "노트북": 1500000,
    "마우스": 30000, 
    "키보드": 50000, 
    "모니터": 300000
}

def checkout(**kwargs) : 
    total_price = 0

    for item, quantity in kwargs.items() :
        if item in inventory :
            item_price = inventory[item] * quantity
            total_price += item_price

            print(f"{item} {quantity}개 : {item_price}원")

        else :
            print(f"판매하지 않는 상품입니다 : {item}")
    print(f"총 결제 금액 : {total_price}")

checkout(노트북 = 1, 마우스 = 2)
print("==============")
checkout(모니터 = 2, 패드 = 1)


```
