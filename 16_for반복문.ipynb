```python
# for, range 사용하기

# for 변수 in range(횟수) :
    # 반복할 코드

for a in range(10) :
    print("hello world!")
```

```python
# 반복문에서 변수의 변화 확인

for i in range(10) :
    print("hello", i)
```

```python
for num in range(5) :
    print("hello", num)
```

```python
# 시작하는 숫자와 끝나는 숫자 지정하기

for a in range(5, 12) :
    print(a)
```

```python
# 증가폭 사용하기

for b in range(0, 10, 2) :
    print(b)
```

```python
# 숫자 감소시키기

for c in range(10, 0, -2):
    print(c)
```

```python
for d in reversed(range(10)) :
    print(d, d)
```

```python
# 입력한 횟수대로 반복하기
count = int(input())

for i in range(count) :
    print("hello world!", i)
```

```python
# 시퀀스 객체로 반복하기

a = [10, 20, 30, 40, 50]

for b in a :
    print(b)
```

```python
fruits = ("orange", "apple", "grape")

for fruit in fruits :
    print(fruit)
```

```python
for letter in 'python' :
    print(letter, end = " ")
```

```python
for letter in reversed('python') :
    print(letter, end = "")
```

```python
# 연습문제 16.5
# p.193

x = [49, -17, 25, 102, 8, 62, 21]

for a in x :
    print(a * 10, end = " ")
```

```python
# 심사문제 16.6

p = int(input())


for h in range(1,10) :
    print(p, "*", h, "=", p * h)   # print(f"{p} * {h} = {p*h}")
```
