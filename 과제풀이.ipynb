```python
distance = float(input("거리를 입력하세요 (km): "))
hour = int(input("현재 시간을 입력하세요 (0-23): "))
price = 3000

if distance > 2 :
    price += 1000 * (distance -2)
    
if hour >= 22 or hour <= 4 :
    price = price * 1.2 # price *= 1.2
    
print(f"요금 : {int(price)}원")
```
