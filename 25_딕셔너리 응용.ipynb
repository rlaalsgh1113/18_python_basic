```python
# 딕셔너리 표현식 사용하기
# {키:값 for 키, 값 in 딕셔너리}

keys = ["a", "b", "c", "d"]
imsi = dict.fromkeys(keys)
print(imsi)
x = {key : value for key, value in imsi.items()}
print(x)
```

```python
{key : 0 for key in dict.fromkeys(["a", "b", "c", "d"]).keys()}
```

```python
{value : 0 for value in {"a" : 10, "b" : 20, "c" : 30, "d" : 40}.values()}
```

```python
# 딕셔너리 표현식에서 if 조건문 사용하기

x = {
    "a" : 10,
    "b" : 20,
    "c" : 30,
    "d" : 40
}

for key, value in x.items() :  # 딕셔너리의 크기가 달라짐
    if value == 20 :
        del x[key]
print(x)
```

```python
x = {
    "a" : 10,
    "b" : 20,
    "c" : 30,
    "d" : 40
}

del x["b"]
print(x)
```

```python
x = {
    "a" : 10,
    "b" : 20,
    "c" : 30,
    "d" : 40
}
y = x.copy()

for key, value in y.items() :
    if value == 20 :
        del x[key]
print(x)
```

```python
x = {
    "a" : 10,
    "b" : 20,
    "c" : 30,
    "d" : 40
}

print({key : value for key, value in x.items() if value != 20}) # 지운게 아니라 필터링을한거라 가능
print(x)
```
