```python
school = {
    '1반': {
        '김철수': {'국어': 85, '영어': 90, '수학': 78},
        '이영희': {'국어': 92, '영어': 88, '수학': 95}
    },
    '2반': {
        '박민수': {'국어': 88, '영어': 85, '수학': 90},
        '최지영': {'국어': 95, '영어': 92, '수학': 88}
    }
}

chulsoo_math = school['1반']['김철수']['수학']
print(f"김철수의 수학 점수: {chulsoo_math}")

minsu_scores = school['2반']['박민수']
minsu_avg = sum(minsu_scores.values()) / len(minsu_scores)
print(f"박민수의 평균 점수: {minsu_avg : .0f}")
```
