```python
# while 반복문 사용하기

# while문 구조

# 초기식
# while 조건식 :
#     반복할 코드
#     변화식

i = 0
while i < 10 :
    print('hello world')
    i += 1
```

```python
# 초깃값 1부터 시작하기

i = 1

while i <= 5 :
    print("world", i)
    i += 1
```

```python
# 초깃값 감소시키기
i = 5

while i > 0 :
    print("hello", i)
    i -= 1
```

```python
# 입력한 값만큼 반복시키기

count = int(input())
i = 0
while i < count :
    print('world!', i)
    i += 1
```

```python
count = int(input())
i = 0
while i < count :
    print('world!', i)
    # i += 1  없으면 무한루프에 빠짐
```

```python
# 연습문제 17.5

i = 2
j = 5

while i <= 32 or j >= 1 :
    print(i,j)
    i *= 2
    j -= 1
```

```python
# 심사문제 17.6

money = int(input())
cost = 1350

while money >= 0 :
    money -= cost
    if money > 0 :
        print(money)
    elif money == 0 :
        print(0)
```
